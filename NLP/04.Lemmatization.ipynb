### Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after lemmatization is called 'lemma', which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example -

```python
## Classification Problem
## Comments of product is a positive review or negative review
## Review ---> eating, eat, eaten [going, gone, goes] ----> go

words=["eating", "eat", "eaten", "going", "gone", "goes","writing", "written", "wrote", "writes", "sportingly", "fairly", "Programming", "Programming", "Program", "Programs", "Programmer", "Programmers", "history", "finally", "final", "finals", "finalist", "finalists", "finale", "finality", "finalization", "finalize", "finalized", "finalizing", "finality", "finally"]
```

### PorterStemmer

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''
lemmatizer.lemmatize("going", pos="v")
```

```python
for word in words:
    print(word+ "---->"+lemmatizer.lemmatize(word, pos="n"))
```

```python
lemmatizer.lemmatize("goes", pos="v")
```

```python
lemmatizer.lemmatize("fairly", pos="v"), lemmatizer.lemmatize("sportingly")
```
