```python
corpus = """Hello Welcome, to the world of Natural Language Processing. This is a simple example to demonstrate how to use NLTK for text processing. NLTK is a powerful library for working with human language data in Python."""
```

```python
corpus
```

```python
print(corpus)
```

```python
## Tokenization
## Sentence --> paragraphs
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization
## Paragraph --> words
## sentence --> words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    # print(sentence)
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
